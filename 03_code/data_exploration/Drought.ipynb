```python
# 1) Load + parse dates
df = pd.read_csv(
    "/Users/ramilmammadov/Desktop/Capstone Project/Datasets/Drought datasets/merged_all.csv"
)
df["MapDate"] = pd.to_datetime(
    df["MapDate"].astype(str), format="%Y%m%d", errors="coerce"
)
df = df.dropna(subset=["MapDate"]).copy()

# Keep FIPS padded (e.g., 02013)
if "FIPS" in df.columns:
    df["FIPS"] = df["FIPS"].astype(str).str.zfill(5)

# County-level keys to group by (order matters for tidy sorting)
group_cols = [c for c in ["FIPS", "County", "State"] if c in df.columns]
if not group_cols:
    raise ValueError(
        "Expected at least one of: FIPS, County, State to identify counties."
    )

# Optional: restrict each year to a fixed window (Jan 07..Oct 21). Leave commented if not needed.
# m = df["MapDate"].dt.month; d = df["MapDate"].dt.day
# df = df[((m > 1) | ((m == 1) & (d >= 7))) & ((m < 10) | ((m == 10) & (d <= 21)))].copy()

# 2) Stage 1 — collapse duplicates per (county, week) so each week counts once
non_agg_cols = set(group_cols + ["MapDate"])
numeric_cols_all = [
    c
    for c in df.columns
    if c not in non_agg_cols and pd.api.types.is_numeric_dtype(df[c])
]

weekly_all_numeric = (
    df.groupby(group_cols + ["MapDate"], dropna=False)[numeric_cols_all]
    .mean()
    .reset_index()
)

# 3) Stage 2 — annualize: average the weekly values within each (county, year)
weekly_all_numeric["Year"] = weekly_all_numeric["MapDate"].dt.year
aggs = {c: "mean" for c in numeric_cols_all}
aggs["MapDate"] = "nunique"  # how many weekly maps contributed

annual_all_numeric = (
    weekly_all_numeric.groupby(group_cols + ["Year"], dropna=False)
    .agg(aggs)
    .rename(columns={"MapDate": "n_weeks"})
    .reset_index()
    .sort_values(group_cols + ["Year"])
)

# 4) Save ONLY this one CSV to the current folder
annual_all_numeric.to_csv("drought_annual_means_by_county_all_numeric.csv", index=False)
print("Saved: drought_annual_means_by_county_all_numeric.csv")

# (optional) quick peek
print("Shape:", annual_all_numeric.shape)
print(annual_all_numeric.head())
```

```python
# use your DataFrame name; here I assume annual_all_numeric
df = annual_all_numeric.copy()

d_cols = [c for c in ["D0", "D1", "D2", "D3", "D4"] if c in df.columns]
for c in d_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df["Dry"] = df[d_cols].sum(axis=1, min_count=1).round(3)

# drop original D* columns
df = df.drop(columns=d_cols)
df
```

```python
# --- pre-clean: drop any old % variants or suffix duplicates before we rebuild ---
dupe_like = [
    c
    for c in df.columns
    if c in ("Dry_%", "Not_Dry_%")
    or c.startswith("Dry_%_")
    or c.startswith("Not_Dry_%_")
    or (c.startswith("Not_Dry_") and c not in {"Not_Dry", "Not_Dry_%"})
]
df = df.drop(columns=dupe_like, errors="ignore")

# --- ensure 'Dry' exists ---
if "Dry" not in df.columns:
    if "Dry_%" in df.columns:
        df = df.rename(columns={"Dry_%": "Dry"})
    else:
        d_cols = [c for c in ["D0", "D1", "D2", "D3", "D4"] if c in df.columns]
        if d_cols:
            df[d_cols] = df[d_cols].apply(pd.to_numeric, errors="coerce")
            df["Dry"] = df[d_cols].sum(axis=1, min_count=1).round(3)
            df = df.drop(columns=d_cols)
        else:
            raise KeyError("Neither 'Dry' nor D0–D4 present to build 'Dry'.")
```

```python
# --- build Not_Dry from 'None' (or infer if missing) ---
if "None" in df.columns:
    df["Not_Dry"] = pd.to_numeric(df["None"], errors="coerce").round(3)
    df = df.drop(columns=["None"])
elif "Not_Dry" not in df.columns:
    df["Not_Dry"] = (100 - df["Dry"]).clip(0, 100).round(3)

# --- rename weeks column to N_Weeks if needed ---
if "n_weeks" in df.columns:
    df = df.rename(columns={"n_weeks": "N_Weeks"})
elif "N_weeks" in df.columns:
    df = df.rename(columns={"N_weeks": "N_Weeks"})

# --- order columns (pre-rename) then rename to % names ---
order = ["Year", "State", "FIPS", "County", "Dry", "Not_Dry", "N_Weeks"]
df = df[
    [c for c in order if c in df.columns] + [c for c in df.columns if c not in order]
]
df = df.rename(columns={"Dry": "Dry_%", "Not_Dry": "Not_Dry_%"})

# --- final safety: remove any exact duplicate column names, keep the latest ---
df = df.loc[:, ~df.columns.duplicated(keep="last")]

# (optional) enforce final order with % names
final_order = ["Year", "State", "FIPS", "County", "Dry_%", "Not_Dry_%", "N_Weeks"]
df = df[final_order + [c for c in df.columns if c not in final_order]]

df.sample(25)
```

```python
df.to_csv("drought_annual_by_county_final.csv", index=False)
```
