```python
INPUT = Path("/Users/ramilmammadov/Desktop/Capstone Project/GitHub Repo/Capstone_Citizensbank_climate-risk_insurance-exposure_monitoring/02_processed_data/wildfire_data.csv")           
OUTPUT = Path("/Users/ramilmammadov/Desktop/Capstone Project/GitHub Repo/Capstone_Citizensbank_climate-risk_insurance-exposure_monitoring/02_processed_data/wildfire_data_preprocessed.csv")

rename_map = {
    "FIPS_CODE": "FIPS",
    "FIRE_YEAR": "Year",
    "STATE": "State",
}

df = pd.read_csv(INPUT)
existing = {old: new for old, new in rename_map.items() if old in df.columns}
missing = [old for old in rename_map if old not in df.columns]

df = df.rename(columns=existing)
df.to_csv(OUTPUT, index=False)

print(f"Saved: {OUTPUT}")
print("Renamed:", ", ".join(f"{k}→{v}" for k, v in existing.items()) or "None")
if missing:
    print("Columns not found in the file:", ", ".join(missing))

df
```
