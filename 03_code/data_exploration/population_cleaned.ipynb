```python
import pandas as pd
```

```python
df = pd.read_csv("../../01_original_data/Financial Data/population_2000_2024/Population by Age and Sex - US, States, Counties.csv")
```

```python
columns_to_keep = ['IBRC_Geo_ID', 'Year', 'Total Population']
df_subset = df[columns_to_keep]
```

```python
df_subset = df_subset[(df_subset["Year"] >= 2000) & (df_subset["Year"] <= 2020)]
```

```python
df_subset_cleaned = df_subset.drop_duplicates(subset=['IBRC_Geo_ID', 'Year'])
```

```python
df_subset_cleaned = df_subset_cleaned.iloc[21:].reset_index(drop=True)
```
