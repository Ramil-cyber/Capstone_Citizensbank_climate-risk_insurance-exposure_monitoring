```python
import pandas as pd

df = pd.read_csv("../../01_original_data/FinancialData/CAGDP1__ALL_AREAS_2001_2023.csv", encoding="latin1")

df_real = df[df["LineCode"] == 1].drop(columns=["LineCode", "Description", "Unit"])
df_current = df[df["LineCode"] == 3].drop(columns=["LineCode", "Description", "Unit"])
df_index = df[df["LineCode"] == 2].drop(columns=["LineCode", "Description", "Unit"])

df_real_melted = df_real.melt(id_vars=["GeoFIPS", "GeoName"], var_name="Year", value_name="Real_GDP")
df_current_melted = df_current.melt(id_vars=["GeoFIPS", "GeoName"], var_name="Year", value_name="Current_GDP")
df_index_melted = df_index.melt(id_vars=["GeoFIPS", "GeoName"], var_name="Year", value_name="GDP_Index")

df_merged = df_real_melted.merge(df_current_melted, on=["GeoFIPS", "GeoName", "Year"], how="outer")\
                          .merge(df_index_melted, on=["GeoFIPS", "GeoName", "Year"], how="outer")
df_merged = df_merged[df_merged["Year"].str.match(r"^\d{4}$", na=False)]

df_merged["Year"] = df_merged["Year"].astype(int)

df_merged["GeoFIPS"] = df_merged["GeoFIPS"].str.replace('"', '').str.zfill(5)

df_merged = df_merged.iloc[46:].reset_index(drop=True)

df_merged.to_csv("../../01_original_data/FinancialData//GDP/GDP_cleaned.csv", index=False)
```
