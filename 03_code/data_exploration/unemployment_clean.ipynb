```python
import pandas as pd

# FIPS_Code, State, Area_Name, Attribute, Value

df = pd.read_csv("../../01_original_data/FinancialData/Unemployment2023.csv")

df["Year"] = df["Attribute"].str.extract(r'_(\d{4})$')
df["Variable"] = df["Attribute"].str.replace(r'_\d{4}$', '', regex=True)

df_wide = df.pivot_table(
    index=["FIPS_Code", "State", "Area_Name", "Year"],
    columns="Variable",
    values="Value",
    aggfunc="first"
).reset_index()

df_wide.columns.name = None 
df_wide = df_wide.rename(columns={
    "Civilian_labor_force": "Civilian_LF",
    "Unemployed": "Unemployed",
    "Employed": "Employed",
    "Unemployment_rate": "Unemployment_Rate"
})

df_wide = df_wide.drop(columns=[
    "Med_HH_Income_Percent_of_State_Total", 
    "Median_Household_Income", 
    "Metro", 
    "Rural_Urban_Continuum_Code", 
    "Urban_Influence_Code"
])

df_wide = df_wide.iloc[24:].reset_index(drop=True)

df_wide.to_csv("../../01_original_data/FinancialData/Unemployment2023_cleaned.csv", index=False)
```
