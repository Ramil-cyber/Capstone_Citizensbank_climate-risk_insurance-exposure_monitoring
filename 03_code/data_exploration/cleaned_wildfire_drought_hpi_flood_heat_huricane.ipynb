```python
import pandas as pd

df = pd.read_csv(
    "/Users/ramilmammadov/Desktop/Capstone Project/GitHub Repo/Capstone_Citizensbank_climate-risk_insurance-exposure_monitoring/02_processed_data/wildfire_drought_hpi_flood_heat_huricane.csv"
)
df
```

```python
drop_cols = [
    "county_fips5",
    "yr",
    "year_x",
    "Year",
    "County",
    "State_ncei",
    "State_fire",
    "year_y",
]
df = df.drop(columns=[c for c in drop_cols if c in df.columns], errors="ignore")

rename_map = {
    "hpi_yoy": "HPI_CHANGE",
    "CZ_Name": "COUNTY",
    "COUNT": "FLOOD_COUNT",
    "TOTAL_DAMAGE_PROPERTY": "FLOOD_TOTAL_DAMAGE_PROPERTY",
    "TOTAL_DURATION_HOURS": "FLOOD_TOTAL_DURATION_HOURS",
    "heat_index": "HEAT_INDEX",
    "State_abbr": "STATE_ABBR",
    "Annual_Mean_Index": "DROUGHT_INDEX",
    "FIRE_FREQUENCY": "FIRE_COUNT",
    "TOTAL_FIRE_SIZE": "FIRE_TOTAL_FIRE_SIZE",
    "deaths_hurricane": "HURRICANE_DEATHS",
    "injuries_hurricane": "HURRICANE_INJURIES",
    "damage_hurricane": "HURRICANE_DAMAGE",
}
df = df.rename(columns={k: v for k, v in rename_map.items() if k in df.columns})

df["HPI_CHANGE"] = pd.to_numeric(df["HPI_CHANGE"], errors="coerce").round(3)
df
```

```python
df.to_csv(
    "/Users/ramilmammadov/Desktop/Capstone Project/GitHub Repo/Capstone_Citizensbank_climate-risk_insurance-exposure_monitoring/02_processed_data/cleaned_wildfire_drought_hpi_flood_heat_huricane.csv",
    index=False,
)
```
