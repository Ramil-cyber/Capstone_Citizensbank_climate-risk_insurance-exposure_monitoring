```python
# Convert Freddie Mae raw data to Parquet file to reduce size

import pandas as pd

freddie_mae = pd.read_csv(
    "../01_original_data/Freddie_Mae_2025Q1.zip",
    delimiter="|",
    header=None,
    low_memory=False,
)
freddie_mae.to_parquet("../01_original_data/Freddie_Mae_2025Q1.parquet", index=False)
```

```python
# Convert Zillow raw data to Parquet file to reduce size

import pandas as pd

zillow = pd.read_csv(
    "../not_pushing/Zip_zhvi_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.csv",
    low_memory=False,
)
zillow.to_parquet(
    "../01_original_data/zillow/Zip_zhvi_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.parquet",
    index=False,
)
```
