CSV Storm data version merged and saved as parquet files.

```python
import pandas as pd
import os

csv_files = [
    f for f in os.listdir("../not_pushing/storm_data/csvs") if f.endswith(".csv")
]

storm_csv = pd.concat(
    [
        pd.read_csv(f"../not_pushing/storm_data/csvs/{f}", low_memory=False)
        for f in csv_files
    ],
    axis=0,
    ignore_index=True,
)

len(storm_csv)
```

```python
storm_csv.drop(columns=["DAMAGE_CROPS"], inplace=True)
```

```python
storm_csv.head()
# storm_csv.info()
```

```python
fifth = len(storm_csv) // 5
storm_csv.iloc[:fifth, :].to_parquet(
    "../01_original_data/fema_storm_data/storm_data_01.parquet",
    index=False,
)

storm_csv.iloc[fifth : (2 * fifth), :].to_parquet(
    "../01_original_data/fema_storm_data/storm_data_02.parquet", index=False
)

storm_csv.iloc[(2 * fifth) : (3 * fifth), :].to_parquet(
    "../01_original_data/fema_storm_data/storm_data_03.parquet", index=False
)

storm_csv.iloc[(3 * fifth) : (4 * fifth), :].to_parquet(
    "../01_original_data/fema_storm_data/storm_data_04.parquet", index=False
)

storm_csv.iloc[(4 * fifth) :, :].to_parquet(
    "../01_original_data/fema_storm_data/storm_data_05.parquet", index=False
)
```
